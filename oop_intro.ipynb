```python
class Game:
    def __init__(self, name, levels):
        self.name = name
        self.levels = levels

    def start_game(self):
        # check if instance self --> is from Game
        if isinstance(self, Game):
            print(f"Game started {self.name}")
        else:
            raise TypeError("the given object must be form type Game")
padel = Game("padel", 3)
print(padel.name, padel.levels)
padel.start_game()

Game.start_game(e)
```

Break.__dict__ holds __init__ and start_break, shared by every break. bb.day = "Sat" and bb.branch = "Alex" add keys only on bb. getsizeof shows the object and its dictionary growing. getrefcount counts the names that still point at an object, plus the temporary reference used by the call itself.</div></div>

<img src="illustrations/oop-where.svg" alt="The object dictionary holds duration, and the class dictionary holds __init__ and start_break" style="width:100%; border-radius:16px; margin-top:14px;">


```python
class Break:
    def __init__(self, duration=15):
        self.duration = duration

    def start_break(self):
        print(f"Break started with duration {self.duration}")


print(type(Break))
d = Break(60)
print(d.__dict__)
print(Break.__dict__)
```

```python
import sys
print(sys.getsizeof(d))
bb = Break(60)
print(sys.getsizeof(bb))
bb.day = "Sat"
print(sys.getsizeof(bb))
```

```python
print(sys.getsizeof(d.__dict__))

print(sys.getsizeof(bb.__dict__))
```

```python
mm = {}
print(sys.getsizeof(mm))
```

```python
mm["name"]="Abbass"
print(sys.getsizeof(mm))
```

```python
bb.branch ='Alex'
print(bb.__dict__)
print(sys.getsizeof(bb))
print(sys.getsizeof(bb.__dict__))
```

```python
print(sys.getrefcount(bb))
```

```python
class Toy:
    def __init__(self, name):
        self.name = name

q = Toy("Qronfela")
import sys
print(sys.getrefcount(q))
```

```python
qq = q
print(sys.getrefcount(q))
print(id(q))
print(id(qq))

print(q)

```
