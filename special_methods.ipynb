```python
class Student:
    def __init__(self, id, name, grade):
        self.id = id
        self.name = name
        self.grade = grade

    def __str__(self):
        return f"{self.name}"

    # representation --> for developer, debugging
    def __repr__(self):
        return f"Student(id={self.id}, name={self.name}, grade={self.grade})"
s = Student(1, "noha", 100)
print(s)  # print object

print([s]) # call represent function __repr__

```

```python
class Student:
    count = 0
    def __init__(self, id, name, grade):
        self.id = id
        self.name = name
        self.grade = grade
        self.__class__.count += 1


    def __str__(self):
        return f"{self.name}"

    # representation --> for developer, debugging
    def __repr__(self):
        return f"Student(id={self.id}, name={self.name}, grade={self.grade})"

    def __del__(self):
        self.__class__.count -= 1
        print(f"---- object {self.name} deleted, Rest in Peace ")

    def __len__(self):
        return len(self.__dict__) # return no of properties

s = Student(1, "noha", 100)
s2 = Student(2, "noha", 100)
s3 = Student(3, "noha", 100)
print(Student.count)
print(len(s3))
```

```python
# class --> connect to db
# init of object ---> test connection , set credits
# start connection --> call created object
```

```python
class DB:
    def __init__(self, db_name, db_host, db_port):
        self.db_name = db_name
        self.db_host = db_host
        self.db_port = db_port
        self.started = False

    def connect(self):
        if self.started:
            print("connection already started")
        else:
            print("connecting to db")
            self.started = True
            return "connected"



myd = DB("ITI", "192.163.1.2", 5432)
print(myd)
myd.connect()
```

```python
myd.connect()
```

```python
class DB:
    def __init__(self, db_name, db_host, db_port):
        self.db_name = db_name
        self.db_host = db_host
        self.db_port = db_port
        self.started = False

    def __call__(self, *args, **kwargs):
        if self.started:
            print("connection already started")
        else:
            print("connecting to db")
            self.started = True
            return "connected"



myd = DB("ITI", "192.163.1.2", 5432)
myd()  # start connection
myd()
```

```python
"""
    all objects from the same class must have the same attributes / properties

    in c++
    class Student{
        public:
        int id;
        string name;
    }
"""
class Student:
    def __init__(self, id, name, grade):
        self.id = id
        self.name = name
        self.grade = grade


std= Student(1, "noha", 100)
std.city = "Mansoura"

print(std.__dict__)



std2 = Student(2, "noha", 100)
print(std2.__dict__)
```

- if you need to prevent adding extra attributes to the object in the run time use slots

```python
class Student:
    __slots__ = ['id', 'name', 'grade']
    def __init__(self, id, name, grade):
        self.id = id
        self.name = name
        self.grade = grade

print(Student.__dict__)
```

```python
s = Student(1,'noha', 100)
print(s.__dict__)
```
