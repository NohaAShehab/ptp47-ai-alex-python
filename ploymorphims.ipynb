```python
import functools

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary



    def print_emp(self):
        print(f"Employee name is  {self.name}")



    def print_emp(self, msg):
        print(f"Employee name is  {self.name}, {msg}")


    def print_emp(self, num ):
        print(f"Employee name is  {self.name}, {num * self.salary}")





```

```python
from functools import singledispatchmethod

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


    @singledispatchmethod
    def print_emp(self, value ):
        # this is default case, values must be int, str , bool
        # raise error
        print(f"Employee name is  {self.name}")


    @print_emp.register
    def _(self, value : str):
        print(f"Employee name is  {self.name}, {value}")

    @print_emp.register
    def _(self, value : int):
        print(f"Employee name is  {self.name}, {value * self.salary}")

    @print_emp.register
    def _(self, value : bool , msg : str):
        print(f"Employee name is  {self.name}, {msg} {value}")


```

```python
class Test:
    @singledispatchmethod
    def start_test(self, name , max_grade ):
        print(f"name is {name}, max_grade is {max_grade}")

    @start_test.register
    def _(self, name :str , max_grade : int):
        print(f"name is {name}, max_grade is {max_grade}, name is str , grade is int ")

    @start_test.register
    def _(self, name : list , max_grade : float):
        print(f"name is {name} list, max_grade is {max_grade} float")


moaz =  Test()
moaz.start_test("python", 20)
moaz.start_test(["python", "ML", "DL"], 520.2)
```

```python
# we all implicitly inherit from object class
# Operator overloading
class Car:
    def __init__(self, brand, color, price):
        self.brand = brand
        self.color = color
        self.price = price
    def __lt__(self, other):
        return self.price < other.price

    def __add__(self, other):
        return self.price + other.price

    def __eq__(self, other):
        return self.price == other.price

    def __iter__(self):
        for item in self.__dict__:
            yield item, self.__dict__[item]




bm = Car("BM", "Red", 5000)
ff = Car("Ferarri", "Red", 5000)
cupra = Car("Cupra", "Red", 6000)

print(bm < cupra)

for item in bm:
    print(item)
```

Defining <b>__call__</b> makes <b>c()</b> and <b>c(34, color="red")</b> land in that method, with the number in <b>args</b> and the color in <b>kwargs</b>.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">say_hi()</span><span style="color:#0f766e;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">c() TypeError</span><span style="color:#0f766e;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">def __call__</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">__call__</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">*args</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">**kwargs</span></div></div>


```python

def say_hi(name):
    print(f"Hi {name}")

say_hi("Noha")  # object form class functions
```

```python
class Car:
    pass

c = Car()

c()
```

```python
class Car:
    def __call__(self, *args, **kwargs):
        print(f"---- My car is called with {args}, {kwargs}")

c = Car()

c()
```

```python
c(34,color='red')
```

<div style="background:#fffbeb;border:1px solid #b45309;border-left:8px solid #b45309;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#b45309;">WITH</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Enter, body, exit</div><div style="line-height:1.55;font-size:15px;"><b>with t</b> calls <b>__enter__</b> before the body and <b>__exit__</b> after it. The exit method receives three values: exception type, exception value, and traceback. They are <b>None</b> when the body finishes cleanly, which is what the first timer prints. The next timer declares only <b>exc_type</b>, so the extra two arguments blow up. <b>*args</b> on the last timer accepts all three and prints them.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#b45309;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">__enter__</span><span style="color:#b45309;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#b45309;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">with body</span><span style="color:#b45309;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#b45309;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">__exit__(exc_type, exc_val, exc_tb)</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">context manager</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">*args</span></div></div>


```python
class Timer:
    def __enter__(self):
        print("----- timer started -----")


    def __exit__(self, exc_type, exc_val, exc_tb):
        print(f"----- timer exited -----  {exc_type, exc_val, exc_tb}")


t = Timer()

with t:
    print("----- examining enter, exit  -----")

```

```python
class Timer:
    def __enter__(self):
        print("----- timer started -----")


    def __exit__(self, exc_type):
        print(f"----- timer exited -----  {exc_type}")


t = Timer()

with t:
    print("----- examining enter, exit  -----")

```

```python
class Timer:
    def __enter__(self):
        print("----- timer started -----")


    def __exit__(self, * args):
        print(f"----- timer exited -----  {args}")


t = Timer()

with t:
    print("----- examining enter, exit  -----")

```
