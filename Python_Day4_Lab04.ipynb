Write a class `Temperature` that stores `celsius`, with one method of each kind:

- `is_valid(celsius)`, a **static** method: `True` if the value is at least `-273.15` (absolute zero).
- `__init__(self, celsius)` raises `ValueError` with `<value> is below absolute zero` when `is_valid` is `False`.
- `to_fahrenheit(self)`, an **instance** method: `celsius * 9 / 5 + 32`, rounded to 1 decimal.
- `from_fahrenheit(cls, f)`, a **class** method: a factory that converts `(f - 32) * 5 / 9`, rounded to 1 decimal, and returns a new object.

```text
77.0
37.0 Temperature
False
```

<div style="background-color:#E3EEFB; border-left:6px solid #2468B8; border-radius:6px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#2468B8;">Note:</b> the cell after your code is a <b>test cell</b>.
````python
class Temperature:
    # A6: a static method is_valid, __init__, an instance method to_fahrenheit,
    #     and a class method from_fahrenheit
    pass


print(Temperature(25).to_fahrenheit())
body = Temperature.from_fahrenheit(98.6)
print(body.celsius, type(body).__name__)
print(Temperature.is_valid(-300))
````

````python
assert Temperature(100).to_fahrenheit() == 212.0
assert Temperature.from_fahrenheit(32).celsius == 0.0
assert isinstance(Temperature.from_fahrenheit(50), Temperature)
assert Temperature.is_valid(-273.15) and not Temperature.is_valid(-274)
try:
    Temperature(-300)
    raise AssertionError("Temperature(-300) must raise ValueError")
except ValueError as e:
    assert str(e) == "-300 is below absolute zero", e
print("A6: all tests passed")
````

````python
class TaskList:
    def __init__(self):
        self.tasks = []

    def add(self, task):
        self.tasks.append(task)

    def is_empty(self):
        return len(self.tasks) == 0

    def next_task(self):
        if self.is_empty:
            return "Nothing to do"
        return self.tasks.pop(0)


todo = TaskList()
todo.add("Write tests")
todo.add("Fix bugs")
print(todo.next_task())
print(todo.next_task())
print(todo.next_task())
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

    @classmethod
    def from_text(cls, text):
        return Animal(text.strip().title())


class Cat(Animal):
    def speak(self):
        return f"{self.name} says meow"


pet = Cat.from_text("  mimi ")
print(type(pet).__name__, "-", pet.speak())
````
