Each object owns a separate count, so the numbers never meet.</div></div>

<img src="illustrations/cv-instance.svg" alt="Two students, s and s2, each running self.count from 0 to 1" style="width:100%; border-radius:16px; margin-top:14px;">


```python
class Student:
    def __init__(self,id,name,age):
        self.id = id
        self.name = name
        self.age = age
        self.count = 0
        self.count +=1

s = Student(1,"",18)
print(s.count)

s2 = Student(1,"",18)
print(s2.count)
```

Student.count += 1 and self.__class__.count += 1 both update that one number. s2.count finds it on the class, because count is not inside s.__dict__. Student.__dict__ is where the shared count shows up.</div></div>

<img src="illustrations/cv-shared.svg" alt="Student.count is 2 on the class. s.__dict__ has id, name, and age, and s2.count looks up to the class" style="width:100%; border-radius:16px; margin-top:14px;">


```python
class Student:
    count = 0 # class variable , shared property between all instances
    def __init__(self,id,name,age):
        self.id = id
        self.name = name
        self.age = age
        Student.count +=1

# class variable access via class
print(Student.count)
s = Student(1,"",18)
s2 = Student(1,"",18)
print(Student.count)
```

```python
print(s2.count)
```

```python
print(s.__dict__)
```

```python
class Student:
    count = 0 # class variable , shared property between all instances
    def __init__(self,id,name,age):
        self.id = id
        self.name = name
        self.age = age
        # Student.count +=1
        self.__class__.count +=1

print(Student.count)
s = Student(1,"",18)
s2 = Student(1,"",18)
print(Student.count)
```

```python
# class method --> define bahaviour class can do ??

class Student:
    registry = {}
    def __init__(self,name):
        self.name = name
        self.__class__.registry[name] = self

    @classmethod # the first parm of the function refers to the class not the objects
    def load_students(cls):
        print(cls, "here")
        return cls.registry

    @classmethod
    def search_student(cls,name):
        return cls.registry.get(name)

s = Student("noha")
print(Student.registry)
print(Student.load_students())

print(Student.search_student("Ahmed"))
print(Student.search_student("noha"))
```

<img src="illustrations/cv-static.svg" alt="cal_net_sal multiplies a salary by 0.8, callable as a function, on emp, or on Employee" style="width:100%; border-radius:16px; margin-top:14px;">


```python
# in python are helper methods

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

emp = Employee("noha", 10000)
def cal_net_sal(salary):
    return salary*.8

print(cal_net_sal(emp.salary))
```

```python
print(cal_net_sal(545572121212))
```

```python
# in python are helper methods

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @staticmethod ##this function is helper method
    def cal_net_sal(salary):
        return salary*.8

emp = Employee("noha", 10000)

emp.cal_net_sal(emp.salary)

Employee.cal_net_sal(545454)


```
