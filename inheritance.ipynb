<div style="background:#f5f3ff;border:1px solid #7c3aed;border-left:8px solid #7c3aed;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#7c3aed;">INHERITANCE</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">The child writes the only constructor</div><div style="line-height:1.55;font-size:15px;">C lists two parents, A and B. C also writes its own <b>__init__</b> and never calls a parent. Creating <b>c = C()</b> runs that one method, so the only line you see is <b>C object is created</b>.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#7c3aed;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">A</span><span style="color:#7c3aed;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#7c3aed;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">B</span><span style="color:#7c3aed;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#7c3aed;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">C(A, B) — only this __init__ runs</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">multiple inheritance</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">no super()</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">C()</span></div></div>


```python
class A:
    def __init__(self):
        print("A object is created")

class B:
    def __init__(self):
        print("B object is created")

class C(A,B):
    def __init__(self):
        print("C object is created")


c = C()
```

<div style="background:#eff6ff;border:1px solid #2563eb;border-left:8px solid #2563eb;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#2563eb;">SUPER · MRO</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">super() walks to the next class</div><div style="line-height:1.55;font-size:15px;">The method resolution order of <b>C(A, B)</b> is C, then A, then B, then object. <b>super().__init__()</b> starts at the next class after C, so A’s constructor runs. B stays quiet.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#2563eb;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">C</span><span style="color:#2563eb;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#2563eb;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">A  ← super() lands here</span><span style="color:#2563eb;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#2563eb;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">B</span><span style="color:#2563eb;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#2563eb;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">object</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">super()</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">MRO</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">A object is created</span></div></div>


```python
class A:
    def __init__(self):
        print("A object is created")

class B:
    def __init__(self):
        print("B object is created")

class C(A,B):
    def __init__(self):
        super().__init__()
        print("C object is created")


c = C()
```

<div style="background:#f0fdfa;border:1px solid #0f766e;border-left:8px solid #0f766e;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#0f766e;">EXPLICIT PARENT</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Call B by name</div><div style="line-height:1.55;font-size:15px;"><b>B.__init__(self)</b> does not consult the MRO. It builds a B directly, so the print comes from B, then from C. A is skipped on purpose.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">C</span><span style="color:#0f766e;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">B.__init__(self)</span><span style="color:#0f766e;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#0f766e;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">B object is created</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">B.__init__(self)</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">skip A</span></div></div>


```python
class A:
    def __init__(self):
        print("A object is created")

class B:
    def __init__(self):
        print("B object is created")

class C(A,B):
    def __init__(self):
        # super().__init__()
        B.__init__(self) # call B constructor
        print("C object is created")


c = C()
```

<div style="background:#fff7ed;border:1px solid #c2410c;border-left:8px solid #c2410c;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#c2410c;">BOTH PARENTS</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Build A and B, then use both methods</div><div style="line-height:1.55;font-size:15px;"><b>super().__init__()</b> still builds A. The next line, <b>B.__init__(self)</b>, builds B. Methods are reached the same way: <b>A.say_hello(self)</b> and <b>B.say_hi(self)</b>.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">super() → A</span><span style="color:#c2410c;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">B.__init__(self)</span><span style="color:#c2410c;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">C.print_test()</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">say_hello</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">say_hi</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">two parents</span></div></div>


```python
class A:
    def __init__(self):
        print("A object is created")

    def say_hello(self):
        print("Hello A")

class B:
    def __init__(self):
        print("B object is created")
    def say_hi(self):
        print("Hi B")

class C(A,B):
    def __init__(self):
        super().__init__()  # A
        B.__init__(self) # call B constructor
        print("C object is created")

    def print_test(self):
        A.say_hello(self)
        B.say_hi(self)




c = C()
```

<div style="background:#f0fdf4;border:1px solid #15803d;border-left:8px solid #15803d;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#15803d;">MULTILEVEL</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Each super() climbs one floor</div><div style="line-height:1.55;font-size:15px;"><b>CEO("noha", 34, 500000, 5)</b> calls Manager, which calls Employee, which calls Person. Every floor keeps the arguments it owns: name, then age, then salary, then level.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#15803d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Person</span><span style="color:#15803d;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#15803d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Employee</span><span style="color:#15803d;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#15803d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Manager</span><span style="color:#15803d;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#15803d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">CEO</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">name</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">age</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">salary</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">level</span></div></div>


```python
class Person:
    def __init__(self, name=""):
        self.name = name

class Employee(Person):
    def __init__(self, name="", age=0):
        super().__init__(name)
        self.age = age

class Manager(Employee):
    def __init__(self, name="", age=0, salary=0):
        super().__init__(name, age)
        self.salary = salary


class CEO(Manager):
    def __init__(self, name=0, age=0, salary=0, level=0):
        super().__init__(name, age,salary)
        self.level = level

noha_ceo = CEO("noha", 34, 500000, 5)



```

<div style="background:#fdf2f8;border:1px solid #be185d;border-left:8px solid #be185d;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#be185d;">COMPOSITION</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">A car has an engine</div><div style="line-height:1.55;font-size:15px;">The first <b>class Car(Engine)</b> says a car <i>is</i> an engine. The second <b>class Car</b> replaces that definition. <b>self.motor = Engine(1, "Gas")</b> says a car <i>has</i> an engine. That second shape is composition: the car holds an engine object.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#be185d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Car</span><span style="color:#be185d;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#be185d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">has a</span><span style="color:#be185d;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#be185d;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Engine(hp, fuel)</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">is-a</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">has-a</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">self.motor</span></div></div>


```python
class Engine:
    def __init__(self, hp, fuel):
        self.hp = hp
        self.fuel = fuel

class Car(Engine):
    pass


# car has engine ....
class Car:
    def __init__(self, model):
        self.model = model
        self.motor = Engine(1,"Gas")  # composition







```
