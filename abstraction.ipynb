```python
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def hello(self): ...

class B:
    def __init__(self):
        print("---Hello from B")


class C(A, B):
    pass


print(C.__mro__)
```

```python
from abc import ABC, abstractmethod

class A(ABC):
    city = "Alex"
    @abstractmethod
    def hello(self): ...

class B:
    def __init__(self):
        print("---Hello from B")


class C(A, B):
    pass


print(C.__mro__)

print(C.city)
```

<div style="background:#fff7ed;border:1px solid #c2410c;border-left:8px solid #c2410c;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#c2410c;">DANGER</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">An abstract class can still run __init__</div><div style="line-height:1.55;font-size:15px;"><b>Do_Not_Do_That</b> cannot be constructed, because <b>help_us</b> is abstract. <b>Errorrs</b> implements <b>help_us</b>, so it can. Its <b>super().__init__()</b> still runs the parent constructor, which prints <b>Errorrrrrrrrr</b> before the child prints anything.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">abstract __init__</span><span style="color:#c2410c;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">concrete Errorrs</span><span style="color:#c2410c;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#c2410c;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">super() runs the print</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">constructor side effect</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">help_us</span></div></div>


```python
class Do_Not_Do_That(ABC):
    def __init__(self):
        self.name =  'Moaz'
        print("Errorrrrrrrrr")

    @abstractmethod
    def help_us(self):pass

```

```python
class Errorrs(Do_Not_Do_That):
    def __init__(self):
        super().__init__()
        print("----- Doing more errors -----")

    def help_us(self):
        print("Sawatooo ya gama3a yemkn 7ad Yengednaaaaaaa")

moaz_object = Errorrs()

moaz_object.help_us()
```

<div style="background:#ecfdf5;border:1px solid #047857;border-left:8px solid #047857;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#047857;">OPEN / CLOSED · DEPENDENCY INVERSION</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Alert talks to a notifier, not to SMS</div><div style="line-height:1.55;font-size:15px;"><b>Alert</b> stores whatever notifier it is given and calls <b>send</b>. SMS and Email are interchangeable. Adding a new channel means a new class, while <b>Alert</b> stays as it is. This first version uses <b>raise NotImplementedError</b> inside a normal method: <b>Notifier()</b> can still be created, and the error appears only when <b>send</b> is called.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#047857;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Alert</span><span style="color:#047857;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#047857;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Notifier</span><span style="color:#047857;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#047857;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">SMS or Email</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">open/closed</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">dependency inversion</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">NotImplementedError</span></div></div>


```python

class Notifier:
    def send(self, msg): raise NotImplementedError("You cannot send base notifier")

class SMS(Notifier):
    def send(self, msg):
        return f"Notification send via sms:::  {msg[:10]}"

class Email(Notifier):
    def send(self, msg):
        return f"Notification send via email {msg}"


class Alert:
    def __init__(self, notifier : Notifier):
        self.notifier = notifier

    def exam_moved(self):
        return self.notifier.send("Exam moved to next week ")


print(Alert(SMS()).exam_moved())

print(Alert(Email()).exam_moved())







```

<div style="background:#ecfdf5;border:1px solid #065f46;border-left:8px solid #065f46;border-radius:16px;padding:16px 18px;color:#0f172a;"><div style="font-size:12px;letter-spacing:.14em;font-weight:700;color:#065f46;">SAME DESIGN, REAL CONTRACT</div><div style="font-size:22px;font-weight:800;margin:4px 0 8px;color:#0f172a;">Notifier becomes an ABC</div><div style="line-height:1.55;font-size:15px;">Same SMS, Email, and Alert. <b>send</b> is now an abstract method, so a half-finished notifier cannot be constructed. <b>Alert(SMS())</b> and <b>Alert(Email())</b> still print the moved-exam message.<div style="margin-top:12px;line-height:2.2;"><span style="display:inline-block;background:#065f46;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Notifier(ABC)</span><span style="color:#065f46;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#065f46;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">SMS.send</span><span style="color:#065f46;font-weight:800;padding:0 6px;">→</span><span style="display:inline-block;background:#065f46;color:white;padding:6px 12px;border-radius:10px;font-weight:700;">Email.send</span></div></div><div style="margin-top:8px;"><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">@abstractmethod</span><span style="display:inline-block;background:#111827;color:#fde68a;padding:2px 8px;border-radius:999px;margin:4px 6px 0 0;font-size:13px;">exam_moved</span></div></div>


```python

class Notifier(ABC):
    @abstractmethod
    def send(self, msg): raise NotImplementedError("You cannot send base notifier")

class SMS(Notifier):
    def send(self, msg):
        return f"Notification send via sms:::  {msg[:10]}"

class Email(Notifier):
    def send(self, msg):
        return f"Notification send via email {msg}"


class Alert:
    def __init__(self, notifier : Notifier):
        self.notifier = notifier

    def exam_moved(self):
        return self.notifier.send("Exam moved to next week ")


print(Alert(SMS()).exam_moved())

print(Alert(Email()).exam_moved())







```
