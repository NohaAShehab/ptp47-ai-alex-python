## Quick reference

| Term | What it is | Written as |
| --- | --- | --- |
| object | a value with a type, an identity and data | `p = Person("Mona")` |
| class | the blueprint that makes objects | `class Person:` |
| `__init__` | runs once on every new object, right after it is created | `def __init__(self, name):` |
| instance variable | data that belongs to one object | `self.name = name` |
| instance method | a function that works on one object | `def greet(self):` · `p.greet()` |
| class variable | data shared by all objects of the class | `count = 0` in the class body · `Person.count` |
| class method | a method that receives the class, not an object | `@classmethod` · `def create(cls, ...):` |

### F2 · A min-max scaler

<div style="background-color:#E1E6EE; border-left:6px solid #1F3A68; border-radius:6px; padding:8px 12px; color:#111111;"><span style="display:inline-block; background-color:#1F3A68; color:#ffffff; border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">F2</span><span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #1F3A68;border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">everything together</span><span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #1F3A68;border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">machine learning</span><span style="display:inline-block; background-color:#B83232; color:#ffffff; border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">Level: Hard</span></div>

Scikit-learn scalers are classes: `fit` learns from data and stores what it learned on the object, `transform` uses it. Write a class `MinMaxScaler`:

- `__init__` sets `min_` and `max_` to `None`;
- `fit(values)` stores the smallest and largest value and returns `self`;
- `transform(values)` returns a list where each value becomes `(v - min_) / (max_ - min_)`, rounded to 2 decimals. Calling it before `fit` raises `RuntimeError("call fit first")`;
- a class method `from_range(low, high)` returns a scaler that is already fitted on that range.

**Expected output**

```text
18 42
[0.0, 0.17, 0.5, 1.0]
[0.33]
[0.25, 0.5, 1.0]
```

<div style="background-color:#E3EEFB; border-left:6px solid #2468B8; border-radius:6px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#2468B8;">Note:</b> the cell after your code is a <b>test cell</b>. Run it after your function: it prints <code>all tests passed</code>, or an <code>AssertionError</code> showing which case failed.</div>

````python
class MinMaxScaler:
    # F2: your code here
    pass


ages = [18, 22, 30, 42]
scaler = MinMaxScaler().fit(ages)
print(scaler.min_, scaler.max_)
print(scaler.transform(ages))
print(scaler.transform([26]))
percent = MinMaxScaler.from_range(0, 100)
print(percent.transform([25, 50, 100]))
````

````python
s = MinMaxScaler()
assert (s.min_, s.max_) == (None, None)
try:
    s.transform([1])
except RuntimeError:
    pass
else:
    raise AssertionError("transform before fit must raise RuntimeError")
assert s.fit([2, 4, 6]) is s, "fit must return self"
assert s.transform([2, 3, 6]) == [0.0, 0.25, 1.0]
r = MinMaxScaler.from_range(10, 20)
assert type(r) is MinMaxScaler and r.transform([15]) == [0.5]
print("F2: all tests passed")
````
