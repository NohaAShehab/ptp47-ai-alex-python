```python
# in-python --> No keyword public, protected, private

# use naming convention to define accessibilty level

"""
    public --> [a-z]
    protected --> _[a-z]
    private --> __[a-z]

"""
```

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary
```

```python
emp = Employee("Ragada", 'r@gamil.com', 1000000)
```

```python
print(emp.name) # public
```

please don't do this, decent engineer keep the rules

```python
print(emp._email)  # Access to a protected member _email of a class
emp._email = 'updated@gmail.com'
```

```python
print(emp.__salary)
```

```python
print(emp.city)
```

```python
print(emp.__dict__)
```
