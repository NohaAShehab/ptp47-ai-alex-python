<p align="center">
<img src="illustrations/instance_data.png" alt="Each Game object keeps its own name and level" width="900">

```python
import json


class Game:
    def __init__(self, name, level):
        """
        Game constructor
        name, value attributes are called instance variables
        as there values depends on caller instance
        :param name: is passed as value to the name attribute
        :param level: is passed as value to the level attribute
        """
        self.name = name
        self.level = level

    def start_game(self, duration):
        print(f"Game {self.name} started with {duration} minutes at level {self.level}")

print(type(Game))
```

```python
g = Game('Mario', 3)
print(g.name)
```

```python
snake = Game('Snake', 3)
print(snake.name)
```

```python
snake.start_game(duration=60)

g.start_game(duration=40)
```

```python
import json
class Laptop:

    count = 0
    made_in = 'China'
    def __init__(self, model, year, color, price):
        self.model = model
        self.year = year
        self.color = color
        self.price = price
        self.__class__.count += 1


    def describe_laptop(self):
        print(f"The laptop is {self.model} with year {self.year} and color {self.color}, price {self.price}")

    @classmethod # used as object factory
    def load_laptops(cls, file_name):
        list_of_laptops = []
        try:
            with open(file_name, 'r') as f:
                data = json.load(f)
                for obj in data:
                    ll = cls(obj['model'], obj['year'], obj['color'], obj['price'])
                    list_of_laptops.append(ll)
        except Exception as e:
            print(e)
            return None
        return list_of_laptops

old_laptops = Laptop.load_laptops("data.json")
print(old_laptops)

```

```python
# == vs is
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

menna = Student("menna", 22)

mon_mon = menna
print(id(menna), id(mon_mon))

print(mon_mon is menna)
print(mon_mon == menna, "here")

menna2 = Student("menna", 22)

print(menna2 is menna)
print(menna2 == menna, "here")
```

```python
# == vs is
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # default action compare object address --> compare address and values
    def __eq__(self, other):
        return self.name == other.name and self.age == other.age

menna = Student("menna", 22)

mon_mon = menna
print(id(menna), id(mon_mon))

print(mon_mon is menna)
print(mon_mon == menna, "here")

menna2 = Student("menna", 22)

print(menna2 is menna)
print(menna2 == menna, "here")
```
